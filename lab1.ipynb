#6.	Reshaping arrays with NumPy:
#Create a function that takes a NumPy array of shape (length,width,height) and converts it in to a vector of shape (length*width*height,1). Use the function array.reshape() for this.
import numpy as np

```python
 

def reshape_array(input_array):
    length , width, height = input_array.shape
    return input_array.reshape(length * width * height, 1)

# example 
input_array = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])  # shape (2, 2, 2)

result = reshape_array(input_array)
print("Original array shape : ", input_array.shape)
print("Reshaped array shape : ", result.shape)
print(result)  
```

Implementing a function on a matrix:
Create a function that takes a matrix as input and returns the softmax (by row) of matrix. The softmax function is defined like this:

softmax(x)=softmax([x_1  x_2… x_n ])=[e^(x_1 )/(∑_j▒e^(x_j ) )    e^(x_2 )/(∑_j▒e^(x_j ) )  …   e^(x_n )/(∑_j▒e^(x_j ) )]

Check if your function is working correctly by using suitable inputs.



```python
import numpy as np

def softmax(x):
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)

#test the softmax function
matrix = np.array([[1, 2, 3],
                [4, 5, 6]])

result = softmax(matrix)

print("Input matrix:\n", matrix)
print("Softmax result:\n", result)
```

9.	Finding dot product using NumPy:

a.	Create a function that implements dot product of two vectors. The input to the function should be two standard python lists. Identify the time taken to evaluate the dot product using a particular example of your choice.


```python
import time
def dot_product_of_matrices(A, B):
    result = 0

    for i in range(len(A)):
        result += A[i] * B[i]

    return result

#example vectors
a = [1, 2, 3, 4, 5]
b = [6, 7, 8, 9, 10]

# Measure execution time
start_time = time.time()

result = dot_product_of_matrices(a, b)

end_time = time.time()

print("Dot Product:", result)
print("Time taken:", end_time - start_time, "seconds")
```

b.	Now create another function that implements dot product of two vectors using np.dot() function. Identify the time taken to evaluate this dot product and compare it with the time taken in part a.

10.	Finding outer product using NumPy:

a)	Create a function that implements outer product of two vectors. The input to the function should be two standard python lists. Identify the time taken to evaluate the outer product using a particular example of your choice.


```python
import time

def outer_product_of_matrices(a, b):
    result = []

    for i in range(len(a)):
        row = []
        for j in range(len(b)):
            row.append(a[i] * b[j])
        result.append(row)

    return result

# example vectors
a = [1, 2, 3]
b = [4, 5]

start_time = time.time()
result = outer_product_of_matrices(a, b)
end_time = time.time()  

print("Outer Product:", result)
print("Time taken:", end_time - start_time, "seconds")
```

Task 11 — L1 and L2 Loss Functions

11(a) L1 Loss using Python lists


```python
def l1_regularization_python(y_true, y_pred):
    loss = 0

    for i in range(len(y_true)):
        loss += abs(y_true[i] - y_pred[i])

    return loss

# example
y_true = [1, 2, 3, 4]
y_pred = [2, 3, 4, 5]
loss = l1_regularization_python(y_true, y_pred)
print("L1 Regularization Loss:", loss)
```

11(b) L1 Loss using NumPy

```python
import numpy as np
def l1_regularization_numpy(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    loss = np.sum(np.abs(y_true - y_pred))
    return loss

# example
y_true = [1, 2, 3, 4]
y_pred = [2, 3, 4, 5]
loss = l1_regularization_numpy(y_true, y_pred)  
print("L1 Regularization Loss:", loss)
```

11(c) L2 Loss using Python lists

```python
def l2_regularization_python(y_true, y_pred):
    loss = 0

    for i in range(len(y_true)):
        loss += (y_true[i] - y_pred[i]) ** 2

    return loss
 #example
y_true = [1, 2, 3, 4]
y_pred = [2, 3, 4, 5]       

loss = l2_regularization_python(y_true, y_pred)
print("L2 Regularization Loss:", loss)
```

11(d) L2 Loss using NumPy

```python
def l2_regularization_numpy(y_true, y_pred):
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    loss = np.sum((y_true - y_pred) ** 2)
    return loss

# example
y_true = [1, 2, 3, 4]
y_pred = [2, 3, 4, 5]   

loss = l2_regularization_numpy(y_true, y_pred)  
print("L2 Regularization Loss:", loss)
```

Task 12 — Matrix and Matrix Addition

12(a) Using standard Python

```python
def matrix_addition_python(a,b):
    row = len(a)
    cols = len(a[0])

    result = []

    for i in range(row):
        row = []
        for j in range(cols):
            row.append(a[i][j] + b[i][j])
        result.append(row)
    return result

# example matrices 
a = [[1, 2, 3], [4, 5, 6]]
b = [[7, 8, 9], [10, 11, 12]]

result = matrix_addition_python(a, b)
print("Matrix Addition Result:")
for row in result:
    print(row)
```

Task 13 — Matrix and Vector Multiplication

13(a) Using standard Python

```python
def matrix_vector_multiply_python(A, v):
    result = []

    for i in range(len(A)):
        total = 0

        for j in range(len(v)):
            total += A[i][j] * v[j]

        result.append(total)

    return result


# Example matrix
A = [
    [1, 2, 3],
    [4, 5, 6]
]

# Example vector
v = [7, 8, 9]

result = matrix_vector_multiply_python(A, v)

print("Matrix:")
print(A)

print("\nVector:")
print(v)

print("\nMatrix × Vector:")
print(result)
```

13(b) Using NumPy

```python
import numpy as np

def matrix_vector_multiply_numpy(A, v):
    return np.dot(A, v)


A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

v = np.array([7, 8, 9])

result = matrix_vector_multiply_numpy(A, v)

print("Matrix:")
print(A)

print("\nVector:")
print(v)

print("\nMatrix × Vector:")
print(result)
```

Task 14 — Matrix and Matrix Multiplication

14(a) Using standard Python

```python
def matrix_multiply_python(A, B):
    rows_A = len(A)
    cols_A = len(A[0])
    cols_B = len(B[0])

    result = []

    for i in range(rows_A):
        row = []

        for j in range(cols_B):
            total = 0

            for k in range(cols_A):
                total += A[i][k] * B[k][j]

            row.append(total)

        result.append(row)

    return result


# Example matrices
A = [
    [1, 2],
    [3, 4]
]

B = [
    [5, 6],
    [7, 8]
]

result = matrix_multiply_python(A, B)

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nA × B:")
print(result)
```
